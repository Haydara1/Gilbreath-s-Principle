### Principe de Gilbreath:

Dans ce notebook, nous avons programmé le tour de magie de Paul Curry.

#### Préparation:

Pour exécuter ce tour, vous pouvez suivre les instructions suivantes:

- Munissez-vous d'un jeu de cartes complet (52 cartes).
- En préalable, triez les cartes de sorte que la couleur alterne:
  
  $ r, n, r, n, ...$

- Vous pouvez commencer le tour!

#### Exécution:

Demandez à un spectateur de couper et découper le jeu autant de fois qu'il le souhaite.

Puis, distribuez un tas, de taille quelconque, en posant les cartes sur la table, une par une.

Ensuite, exécuter (ou demander à un spectateur) un [riffle shuffle](https://www.youtube.com/watch?v=CNkhBtOx2ug), avec les deux tas, ce mélange peut être imparfait.

Distribuez carte par carte, en alternant entre deux tas, un nombre pair de carte (une dizaine). 

Soulevez votre tas, et demandez au spectateur de deviner la couleur de votre carte (une par une).

Si le spectateur a eu bon, écrivez correct, sinon incorrect. 

**Partie cachée du tour:** si votre carte est rouge, écrivez un espace à la fin de la réponse.

Par exemple: 

Si votre tas est comme ceci: $ r, r, n, n, r $

Vous devez entrez cela: 

```
correct␣
correct␣
correct
incorrect
incorrect␣
```


````python
# Constante
WHITE_SPACE = " "
ENTER = ""

# Nombre de fois où le spectateur a bien deviné.
correct = 0 
numCards = 0

IsRedCard = []

print("Enter correct if the spectator has got a correct answer.")
print("Enter incorrect if the spectator has got a wrong answer.")
print("Press enter when finished.")

# Partie magicien.
while(True):

    # Lire la réponse du magicien.
    s = input("Card number " + str(numCards + 1) + " :")
    
    # S'arrêter:
    if s == ENTER:
        break

    # Si la carte est rouge, le magicien entrera " " en dernier caractère.
    c = s[-1]

    if(c == WHITE_SPACE):
        IsRedCard.append(False)
        
    else:
        IsRedCard.append(True)

    # c pour correct et i pour incorrect
    c = s[0] 

    if c == 'c':
        correct += 1

    
    numCards += 1

print("You have got " + str(correct) + " out of " + str(numCards) + " cards.")
````

````python
# Partie spectateur:
print("Spectator's turn:")
print("Press r for red and b for black, then press enter")
print("You can lie or tell the truth, the computer already knows the answer")

for i in range(numCards):
    # r for red and b for black
    s = input("Card number " + str(i + 1) + ": ") 

    if (s == 'r' and IsRedCard[i]) or (s == 'b' and not IsRedCard[i]):
        print("You are telling the truth")

    else:
        print("You are lying")
        
````
